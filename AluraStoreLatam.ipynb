### Importación de datos



```python
import pandas as pd

url = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_1%20.csv"
url2 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_2.csv"
url3 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_3.csv"
url4 = "https://raw.githubusercontent.com/alura-es-cursos/challenge1-data-science-latam/refs/heads/main/base-de-datos-challenge1-latam/tienda_4.csv"

tienda1 = pd.read_csv(url)
tienda2 = pd.read_csv(url2)
tienda3 = pd.read_csv(url3)
tienda4 = pd.read_csv(url4)

tienda1.head()
```

```python
tienda1['No. tienda'] = 1
tienda2['No. tienda'] = 2
tienda3['No. tienda'] = 3
tienda4['No. tienda'] = 4
tienda1.head(5)
```

```python
df_tiendas = pd.concat([tienda1, tienda2, tienda3, tienda4])
df_tiendas.groupby('No. tienda').head(2)
```

```python
df_tiendas.info()
```

# ***Análisis de datos de la tienda***

## 1. Análisis de facturación



### Ingreso total por cada tienda.

```python
ingreso_total = df_tiendas.groupby('No. tienda')['Precio'].sum().astype(int).reset_index(name='Precio')
ingreso_total
```

## 2. Ventas por categoría

```python
df_ventas_por_categoria_tienda = df_tiendas.groupby(['No. tienda', 'Categoría del Producto']).size().reset_index(name='Cantidad de Ventas').sort_values(by=['No. tienda','Cantidad de Ventas'])
df_ventas_por_categoria_tienda
```

```python
categorias = df_ventas_por_categoria_tienda['Categoría del Producto'].unique()
categorias
```

```python
tiendas_numeros = df_ventas_por_categoria_tienda['No. tienda'].unique()
tiendas_numeros.sort()
tiendas_numeros
```

## 3. Calificación promedio de la tienda


```python
calificacion_promedio_por_tienda = df_tiendas.groupby('No. tienda')['Calificación'].mean().reset_index(name='Calificación')
calificacion_promedio_por_tienda
```

```python
# Tienda dos
calificacion_promedio_2 = tienda2["Calificación"].mean()
print(f"La calificación promedio de la tienda es: {calificacion_promedio_2}")
```

## 4. Productos más y menos vendidos

```python
df_conteo_productos = df_tiendas.groupby(['No. tienda', 'Producto']).size().reset_index(name='Cantidad de Ventas Producto')
df_conteo_productos
```

```python
df_productos_mas_vendidos = df_conteo_productos.loc[df_conteo_productos.groupby('No. tienda')['Cantidad de Ventas Producto'].idxmax()]
df_productos_menos_vendidos = df_conteo_productos.loc[df_conteo_productos.groupby('No. tienda')['Cantidad de Ventas Producto'].idxmin()]

# Renombrar las columnas para mayor claridad en el resultado final
df_productos_mas_vendidos = df_productos_mas_vendidos.rename(columns={'Producto': 'Producto Más Vendido', 'Cantidad de Ventas Producto': 'Cantidad Más Vendida'})
df_productos_menos_vendidos = df_productos_menos_vendidos.rename(columns={'Producto': 'Producto Menos Vendido', 'Cantidad de Ventas Producto': 'Cantidad Menos Vendida'})

# Combinar los resultados en un solo DataFrame
df_max_min = pd.merge(df_productos_mas_vendidos[['No. tienda', 'Producto Más Vendido', 'Cantidad Más Vendida']],
                      df_productos_menos_vendidos[['No. tienda', 'Producto Menos Vendido', 'Cantidad Menos Vendida']],
                      on='No. tienda')
df_max_min
```

## 5. Envío promedio por tienda

```python
envio_prom_tienda = df_tiendas.groupby('No. tienda')['Costo de envío'].mean().reset_index(name='Costo de envío')
envio_prom_tienda
```

# ***Visualización***

## Generando gráfico

```python
fig, ax = plt.subplots(figsize=(10, 8))

lable = ingreso_total['No. tienda'].replace({1: 'Tienda 1', 2: 'Tienda 2', 3: 'Tienda 3', 4: 'Tienda 4'})
plt.pie(ingreso_total['Precio'], labels=lable, autopct='%1.1f%%')
plt.title('Distribución de Ingresos por Tienda')

plt.legend()
plt.grid(True)

plt.show()
```

```python
#gráfica promedio envios vs ingresos totales por tienda.
fig, ax1 = plt.subplots(figsize=(10, 6))

lable = ingreso_total['No. tienda'].replace({1: 'Tienda 1', 2: 'Tienda 2', 3: 'Tienda 3', 4: 'Tienda 4'})
# 2. Primer eje (Ingresos - Barras)
ax1.set_xlabel(lable)
ax1.set_ylabel('Ingresos Totales ($)', color='navy', fontsize=12)
barras = ax1.bar(lable, ingreso_total['Precio'], color='skyblue', alpha=0.7, label='Ingresos')
ax1.tick_params(axis='y', labelcolor='navy')

# 3. Segundo eje (Costos de Envío - Línea con puntos)
ax2 = ax1.twinx()
ax2.set_ylabel('Promedio Costo de Envío ($)', color='red', fontsize=12)
linea = ax2.plot(lable, envio_prom_tienda['Costo de envío'], color='red', marker='o', linewidth=2, label='Costo Envío Prom.')
ax2.tick_params(axis='y', labelcolor='red')

# 4. Título y detalles
plt.title('Comparativa: Ingresos vs. Promedio Envío por Tienda', fontsize=14)
ax1.grid(axis='y', linestyle='--', alpha=0.5)

plt.show()
```

```python
# Gráfico de calificacion promedio de las tiendas



fig, ax = plt.subplots()
lable = calificacion_promedio_por_tienda['No. tienda'].replace({1: 'Tienda 1', 2: 'Tienda 2', 3: 'Tienda 3', 4: 'Tienda 4'})
ax.plot(lable,calificacion_promedio_por_tienda['Calificación'], color = 'tab:purple')
ax.set_xlabel('Tiendas')
ax.set_ylabel('Calificación')
plt.show()
```

```python

import numpy as np
# Configurar el tamaño de la figura
fig, ax = plt.subplots(figsize=(16, 10))

bar_width = 0.18 # Ancho de cada barra
index = np.arange(len(categorias)) # Posiciones en el eje X para cada categoría

# Definir colores para cada tienda
colors = ['#1f77b4', '#ff7f0e', '#2ca02c', '#d62728'] # Colores distintivos para 4 tiendas

# Plotear las barras para cada tienda
for i, tienda_num in enumerate(tiendas_numeros):
    # Filtrar los datos para la tienda actual
    ventas_tienda = df_ventas_por_categoria_tienda[df_ventas_por_categoria_tienda['No. tienda'] == tienda_num]

    # Crear un diccionario para mapear categoría a cantidad de ventas, llenando con 0 si la categoría no existe para esa tienda
    ventas_por_categoria_dict = ventas_tienda.set_index('Categoría del Producto')['Cantidad de Ventas'].to_dict()
    ventas_data = [ventas_por_categoria_dict.get(cat, 0) for cat in categorias]

    # Posicionar las barras agrupadas
    ax.bar(index + i * bar_width, ventas_data, bar_width, label=f'Tienda {tienda_num}', color=colors[i])

# Añadir etiquetas, título y leyenda
ax.set_xlabel('Categoría del Producto', fontsize=12)
ax.set_ylabel('Cantidad de Ventas', fontsize=12)
ax.set_title('Cantidad de Productos Vendidos por Categoría y Tienda', fontsize=14)
ax.set_xticks(index + bar_width * (len(tiendas_numeros) - 1) / 2) # Centrar las etiquetas X en el grupo de barras
ax.set_xticklabels(categorias, rotation=45, ha='right', fontsize=10) # Rotar etiquetas para mejor lectura
ax.legend(title='No. Tienda', fontsize=10, title_fontsize=12)
ax.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout() # Ajustar el diseño para evitar superposiciones
plt.show()
```

## Introducción

El presente informe tiene como objetivo evaluar el desempeño de las cuatro unidades de negocio de la cadena, utilizandométricas clave de ventas y satisfacción del cliente. A través de este análisis comparativo, se busca identificar la tienda con menor ediciencia relativa para fundamentar una decisión de desinversión , permitiendo así la reasignación de capital hacia un nuevo proyecto empresarial todo esto basado y fundamentado en datos de cada una de la unidades de negocio.

## Desarrollo

Para este análisis, siguiendo la metodología y preparación de datos, se integraron cuatro fuentes de datos independientes en un único repositorio consolidado. Se añadió un identificador único por unidad empresarial, lo que permitió realizar un análisis comparativo estandarizado de ños indicadores de rendimiento entre las diferentes unidades de negocio del Sr. Juan.

Para poder compreder mejor los datos se presenta una serie de gráficas.
* La primer gráfica llamada 'Distribución de ingresos por tienda' representa las ventas o ingresos donde se puede evidenciar la contribución de cada una de las tiendas al ingreso total.

    *   La **Tienda 1** es la que generó el mayor porcentaje de ingresos, contribuyendo con el **26.4%** del total.
    *   La **Tienda 4** es la que generó el menor porcentaje de ingresos, aportando un **23.9%** al total.

    En términos de ingresos, las cuatro tiendas presentan un rendimiento muy similar, con una variación muy mínima entre estas, pues entre la mejor calificada y la peor calificada tiene una variación máxima de **2.5%**. Este dato es crucial: indica que la decisión no puede basarse solo en este indicador, ya que todas las sucursales contribuyen de manera realtivamente equitativa. Por lo tanto, el factor determinante para la desición de desinversión deberá tomarse considerando la eficiencia operativa y/o las reseñas/calificaciones de los clientes.
* La gráfica 'Comparativa:Ingresos vs. Promedio costo envíos por tienda' explica la relación entre los ingresos y los costos de envío promedio de cada tienda.

  **Tendencias y Patrones Observados:**

  *   **Ingresos:** La Tienda 1 presenta los ingresos totales más altos, seguida de la Tienda 2, Tienda 3 y finalmente la Tienda 4 con los ingresos más bajos.
  *   **Costo de Envío Promedio:** Existe una tendencia general a que las tiendas con mayores ingresos tienen también un costo de envío promedio ligeramente más alto. Sin embargo, esta relación no es linealmente directa. La Tienda 1, con los ingresos más altos, tiene el costo de envío promedio más elevado. Curiosamente, la Tienda 4, con los ingresos más bajos, también tiene el costo de envío promedio más bajo, lo que podría indicar una operación más eficiente en costos de envío o un menor volumen de ventas de productos con altos costos de envío.

    ¿La Tienda 4 es barata de operar o simplemente vende poco?

    Cálculo de "Costo de Envío por cada $1M de ingreso":

    T1: $26,018 / 1,150M = 22.6

    T2: $25,216 / 1,116M = 22.5

    T3: $24,805 / 1,098M = 22.5

    T4: $23,459 / 1,038M = 22.5

    Estas operaciones demuestran que los costos de envío son lineales. No hay una "ventaja logística" en la Tienda 4; sus costos son bajos solo porque sus ventas son bajas. Su eficiencia operativa es idéntica a las demás, pero su escala es menor.

* La gráfica "Calificación de las tiendas" proporciona una explicación de loas tendencias de satisfacción del cliente entre las tiendas.
    *   **Tienda con Calificación Más Alta**: La Tienda 3 registra la calificación promedio más alta con aproximadamente **4.048**.
    *   **Tienda con Calificación Más Baja**: La Tienda 1 muestra la calificación promedio más baja con aproximadamente **3.977**.

    Apesar de que todas las tiendas operan bajo un estándar de calidad alto y que la experiencia del cliente es equilibrada a nivel general(promedio de 4.0), las ligeras variaciones cobran valor cuando se cruzan con el rendimiento financiero. La identificación de la tienda menos eficiente se centrará en aquiellas que, aun manteniendo una buena reputación, presentan los márgenes de ingresos más bajos. Teniendo en este punto en miras a la tienda 4 como la menos eficiente.
* La última gráfica "Cantidad de productos vendidos por categoria y tienda"  compara el volumen de ventas de productos por categoría entre las diferentes tiendas.
  * Se puede evidenciar cuales son las categorías con más y menos  ventas en cada uno de los puntos de negocio.
  * La tienda 4 vende en categorías de bajo costo (Hogar/Juguetes), pero no logra capturar el mercado de alto valor (Electrónica/Muebles). Esto explica el porqué sus ingresos totales son los más bajos.
  * Mientras las tiendas 3 y 1 domina en "Muebles" y "Electrónicos" (lo que explica su alta facturación), la Tienda 4 no tiene una categoría "estrella" de alto valor que justifique mantenerla abierta frente a una desinversión.

## Conclusión

Tras cruzar las métricas de rendimiento, se recomienda la venta de la tienda 4 basandose en estos tres pilares:
1. Al ser la tienda con la facturación más baja ($1.038M) su salida es la que menos debilitaría la estructura de ingresos totales de la cadena de establecimientos comerciales.
2. Aunque sus costos de envío son los menores en términos absolutos, el analisis de ratios demuestra que no existe una eficiencia superios; simplemente refleja un menor volumen de actividad.
3. La tienda 4 se encuentra en un "punto medio gris": no tiene la lealtad masiva de la tienda 1 ni la excelencia en servicio de la tienda 3. Es la unidad con menor potencial de retorno estratégico se se decide conservar.

Finalmente, al desglozar las ventas por categoría, se observa que la Tienda 4 presenta dependencia de productos como  'Artículos para el hogar', mientras muestra un rezago significativo en categorías de alto valor como 'Electrónicos' y 'Electrodomésticos'. Esta falta de competitividad en los segmentos más rentables del mercado confirma que la Tienda 4 es la unidad con menor eficiencia comercial, validando la recomendación de desinversión para priorizar el nuevo emprendimiento.
